1. Even or Odd Checker


Write a program that takes a number and prints:
    
    
• "Even" if divisible by 2


• "Odd" otherwise


Concept: if condition

```python
num = int(input("Enter any number:"))

if (num % 2 == 0):
    print("Entered number is Even")
else:
    print("Entered number is Odd")
```

2. Positive, Negative or Zero


Take a number and print:
    
    
• "Positive"


• "Negative"


• "Zero"


Concept: multiple if-elif-else

```python
num = int(input("Enter any number:"))

if (num< 0):
    print("Entered number is Negative")
elif (num>0):
    print("Entered number is Positive")
else:
    print("Entered number is Zero")
```

3. Print Numbers from 1 to N


Take a number N and print numbers from 1 to N using a while loop.


Concept: while loop basics

```python
num = int(input("Enter any number:"))

i=1

while i<=num:
    print(i)
    i= i+1
```

4. Sum of First N Numbers


Input: N
    
    
Output: sum from 1 + 2 + ... + N
    
    
Use while loop

```python
num = int(input("Enter any number:"))
i=1
sum=0

while i <=num:
    sum = sum +i
    i = i+1
print("Sum From 1 to",num,"is:",sum)
```

5. Password Retry System


Ask user to enter password:
    
    
• If correct → print "Access Granted"



• If wrong → ask again



• Stop after 3 attempts



Concept: while + if + counter

```python

password ="Sireesha"
i =1

while i <=3:
    ent_password = input("Enter your Password:")
    if ent_password == password:
        print("Access Granted")
        break
    else:
        print("Try Again")
        i= i+1
```

6. Multiplication Table


Input: a number
    
    
Output: print its table up to 10
    
    
Example:
    
    
5 x 1 = 5


5 x 2 = 10


...
Concept: loop + formatting
    
    

```python
num = int(input("Enter any number:"))
rag =10

for i in range(rag+1):
    print(num,"*",i,"=",num*i)
```
